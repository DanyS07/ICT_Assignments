```python
import numpy as np
import matplotlib.pyplot as plt

# Activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def linear(x):
  return(x)

def relu(x):
    return np.maximum(0, x)

def tanh(x):
    return np.tanh(x)

def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

# Input range
x = np.linspace(-10, 10, 400)

# ----Scalar activation plots ----

# Sigmoid
plt.figure()
plt.plot(x, sigmoid(x))
plt.xlabel("x")
plt.ylabel("Sigmoid(x)")
plt.title("Sigmoid Activation Function")
plt.grid(True)
plt.show()

#Linear
plt.figure()
plt.plot(x, linear(x))
plt.xlabel("x")
plt.ylabel("linear(x)")
plt.title("Linear Activation Function")
plt.grid(True)
plt.show()

# ReLU
plt.figure()
plt.plot(x, relu(x))
plt.xlabel("x")
plt.ylabel("ReLU(x)")
plt.title("ReLU Activation Function")
plt.grid(True)
plt.show()


# Tanh
plt.figure()
plt.plot(x, tanh(x))
plt.xlabel("x")
plt.ylabel("Tanh(x)")
plt.title("Tanh Activation Function")
plt.grid(True)
plt.show()

# Softmax
x_softmax = np.linspace(-5, 5, 10)
y_softmax = softmax(x_softmax)

plt.bar(x_softmax, y_softmax)
plt.title("Softmax Activation Function")
plt.xlabel("x")
plt.ylabel("Softmax(x)")
plt.grid(True)
plt.show()
```
