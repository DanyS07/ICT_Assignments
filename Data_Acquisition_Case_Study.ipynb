## Q.16 In which season (Spring, Summer, Fall, Winter) people rented bikes the most?

```python
import pandas as pd
 
data = pd.read_csv("hour.csv")
 
season_map = {
    1: "Spring",
    2: "Summer",
    3: "Fall",
    4: "Winter"
}

data["season_name"] = data["season"].map(season_map)

 
season_rentals = data.groupby("season_name")["cnt"].sum()

print(season_rentals)


print("Season with highest bike rentals:")
print(season_rentals.idxmax())
```

## Q.17 What is the peak hour in which bike rents the most?

```python
import pandas as pd

data = pd.read_csv("hour.csv")


hourly_rentals = data.groupby("hr")["cnt"].sum()


peak_hour = hourly_rentals.idxmax()
peak_rentals = hourly_rentals.max()

print("Peak hour:", peak_hour)
print("Total rentals at peak hour:", peak_rentals)
```

## Q.18 In which day of a week bikes rents out most?

```python
import pandas as pd

data = pd.read_csv("hour.csv")

weekday_map = {
    0: "Sunday",
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday"
}
data["weekday_name"] = data["weekday"].map(weekday_map)


weekday_rentals = data.groupby("weekday_name")["cnt"].sum()


print("Total bike rentals by weekday:")
print(weekday_rentals)


print("\nDay with highest bike rentals:")
print(weekday_rentals.idxmax())

```

## Q1. Which season has the highest average temperature, and how does rental demand behave in that season?

```python
season_map = {
    1: "Spring",
    2: "Summer",
    3: "Fall",
    4: "Winter"
}
data["season_name"] = data["season"].map(season_map)

avg_temp_season = data.groupby("season_name")["temp"].mean()

print("Average temperature by season:\n")
print(avg_temp_season)


highest_temp_season = avg_temp_season.idxmax()
print("\nSeason with highest average temperature:")
print(highest_temp_season)


avg_rentals_season = data.groupby("season_name")["cnt"].mean()

print("\nAverage bike rentals by season:\n")
print(avg_rentals_season)

print("\nAverage rentals in the hottest season:")
print(avg_rentals_season[highest_temp_season])
```

## Q2. Is bike rental demand higher on weekends or weekdays?

```python
day_type_map = {
    0: "Weekend / Holiday",
    1: "Weekday"
}
data["day_type"] = data["workingday"].map(day_type_map)

avg_rentals_day_type = data.groupby("day_type")["cnt"].mean()

print("Average bike rentals by day type:\n")
print(avg_rentals_day_type)
 
print("\nHigher bike rental demand on:")
print(avg_rentals_day_type.idxmax())

```

## Q3. Which combination of season and hour produces the maximum bike rental count?

```python
season_map = {
    1: "Spring",
    2: "Summer",
    3: "Fall",
    4: "Winter"
}
data["season_name"] = data["season"].map(season_map)

# Total rentals for each season–hour combination
season_hour_rentals = data.groupby(["season_name", "hr"])["cnt"].sum()

# Find the combination with maximum rentals
max_combination = season_hour_rentals.idxmax()
max_value = season_hour_rentals.max()

print("Season–Hour combination with maximum bike rentals:")
print("Season:", max_combination[0])
print("Hour:", max_combination[1])
print("Total rentals:", max_value)
```

## Q4. How does average bike rental count vary across different weather situations?

```python
# Map weather situation codes to descriptions
weather_map = {
    1: "Clear / Few clouds",
    2: "Mist / Cloudy",
    3: "Light Snow / Light Rain",
    4: "Heavy Rain / Snow"
}
data["weather_desc"] = data["weathersit"].map(weather_map)

# Average bike rentals by weather situation
avg_rentals_weather = data.groupby("weather_desc")["cnt"].mean()

print("Average bike rentals across weather situations:\n")
print(avg_rentals_weather)

```

## Q5. Which hour shows the lowest bike rental activity across the entire dataset?

```python
# Total rentals per hour
hourly_rentals = data.groupby("hr")["cnt"].sum()

# Hour with lowest rentals
lowest_hour = hourly_rentals.idxmin()
lowest_value = hourly_rentals.min()

print("\nHour with lowest bike rental activity:")
print("Hour:", lowest_hour)
print("Total rentals:", lowest_value)

```
