```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.preprocessing import LabelEncoder
```

```python
data = pd.read_csv('Social_Network_Ads.csv')
data.head()
```

```python
data.isna().sum()
```

```python
data.drop('User ID', axis=1, inplace=True)
```

```python
label_encoder=LabelEncoder()
for col in data:
    data[col]=label_encoder.fit_transform(data[col])
print("Categorical DataFrame after Label Encoding:")
data.head()
```

```python
X = data.drop('Purchased', axis=1)
y = data['Purchased']
y.head(10)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.30, random_state=42)
y_train.head()
```

```python
scaler= StandardScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns)
X_train = X_train_scaled 
X_train.head()
```

```python
scaler= StandardScaler()
X_test_scaled = pd.DataFrame(scaler.fit_transform(X_test), columns=X_test.columns)
X_test = X_test_scaled 
X_test.head()
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
y_pred_log = model.predict(X_test)

log_accuracy = accuracy_score(y_test, y_pred_log)
log_precision = precision_score(y_test, y_pred_log)
log_recall = recall_score(y_test, y_pred_log)
log_f1 = f1_score(y_test, y_pred_log)

print(f"Accuracy: {log_accuracy}")
print(f"Precision: {log_precision}")
print(f"Recall: {log_recall}")
print(f"F1 Score: {log_f1}")
```

```python
knn_sklearn = KNeighborsClassifier(n_neighbors=7)
knn_sklearn.fit(X_train, y_train)
y_pred_knn = knn_sklearn.predict(X_test)

knn_accuracy = accuracy_score(y_test, y_pred_knn)
knn_precision = precision_score(y_test, y_pred_knn)
knn_recall = recall_score(y_test, y_pred_knn)
knn_f1 = f1_score(y_test, y_pred_knn)

print(f"Accuracy: {knn_accuracy}")
print(f"Precision: {knn_precision}")
print(f"Recall: {knn_recall}")
print(f"F1 Score: {knn_f1}")
```

```python
svmclf = SVC(kernel='linear')
svmclf.fit(X_train, y_train)

y_pred_svmL = svmclf.predict(X_test)
```

```python
svmL_accuracy = accuracy_score(y_test, y_pred_svmL)
svmL_precision = precision_score(y_test, y_pred_svmL)
svmL_recall = recall_score(y_test, y_pred_svmL)
svmL_f1 = f1_score(y_test, y_pred_svmL)
```

```python
print(f"Accuracy: {svmL_accuracy}")
print(f"Precision: {svmL_precision}")
print(f"Recall: {svmL_recall}")
print(f"F1 Score: {svmL_f1}")
```

```python
svmcpf = SVC(kernel='poly')
svmcpf.fit(X_train, y_train)

y_pred_svmP = svmcpf.predict(X_test)
```

```python
svmP_accuracy = accuracy_score(y_test, y_pred_svmP)
svmP_precision = precision_score(y_test, y_pred_svmP)
svmP_recall = recall_score(y_test, y_pred_svmP)
svmP_f1 = f1_score(y_test, y_pred_svmP)
```

```python
print(f"Accuracy: {svmP_accuracy}")
print(f"Precision: {svmP_precision}")
print(f"Recall: {svmP_recall}")
print(f"F1 Score: {svmP_f1}")
```

```python
svmcrbf = SVC(kernel='rbf')
svmcrbf.fit(X_train, y_train)

y_pred_svmrbf = svmcrbf.predict(X_test)
```

```python
svmRBF_accuracy = accuracy_score(y_test, y_pred_svmrbf)
svmRBF_precision = precision_score(y_test, y_pred_svmrbf)
svmRBF_recall = recall_score(y_test, y_pred_svmrbf)
svmRBF_f1 = f1_score(y_test, y_pred_svmrbf)
```

```python
print(f"Accuracy: {svmRBF_accuracy}")
print(f"Precision: {svmRBF_precision}")
print(f"Recall: {svmRBF_recall}")
print(f"F1 Score: {svmRBF_f1}")
```

```python
result_df = pd.DataFrame(columns=["Model", "Test Accuracy", "Test Precision","Test Recall","Test F1"])
result_df.loc[len(result_df)] = ["Logostic Regression", log_accuracy, log_precision, log_recall, log_f1]
result_df.loc[len(result_df)] = ["KNN", knn_accuracy, knn_precision, knn_recall, knn_f1]
result_df.loc[len(result_df)] = ["SVM Linear", svmL_accuracy, svmL_precision, svmL_recall, svmL_f1]
result_df.loc[len(result_df)] = ["SVM Poly", svmP_accuracy, svmP_precision, svmP_recall, svmP_f1]
result_df.loc[len(result_df)] = ["SVM RBF", svmRBF_accuracy, svmRBF_precision, svmRBF_recall, svmRBF_f1]
result_df
```
